# 🏥 Clinical Trial Patient Matching Agent

Thin entry point. The pipeline now lives in modular `.py` files:

| Module (in `ctmatch/`) | Responsibility |
|--------|----------------|
| `config.py` | API key (from `GROQ_API_KEY`), models, `call_llm`, `clean_json` |
| `patients.py` | Sample patients + `ACTIVE_PATIENT` |
| `ehr_extraction.py` | `extract_patient_profile()` |
| `trial_search.py` | `search_clinicaltrials`, `parse_all_trials`, `pre_filter` |
| `criteria_reasoning.py` | `reason_over_all_trials()` |
| `scoring.py` | `score_and_rank()` |
| `display.py` | match cards, missing-data plan, export |
| `pipeline.py` | `run_pipeline()` — runs everything in order |

Prefer the terminal? Use the `ctmatch` command instead (see README).
